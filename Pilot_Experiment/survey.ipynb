## Output format

Follow these rules:

A20 and A20b: exactly one of GO UP, STAY THE SAME, GO DOWN, DON'T KNOW.

````python
persona = {
    "persona_id": "P001",
    "age": 36,
    "gender": "male",
    "marital_status": "married/partner",
    "education": "earned a college degree after further education",
    "income": "$175,000–$199,999",
    "region": "North Central",
    "children": 2,
    "adults": 2,
    "homeownership": "owns or buying",
    "survey_date": "December 2021"
}
````

````python
prompt = f"""

## Persona

You are answering questions as if you were a human.  
Do not break character.

You are a {persona['age']}-year-old {persona['gender']} who is
{persona['marital_status']}, with an education level of
{persona['education']}, an income category of {persona['income']},
and who lives in the region of {persona['region']}.

Your household has {persona['children']} children and
{persona['adults']} adults.
Your housing status is: {persona['homeownership']}.

Your demographic characteristics describe the respondent you are representing. Do not assume that people with these characteristics necessarily share the same opinions or expectations. Answer as an individual.

## Knowledge restriction

You are responding to this survey in {persona['survey_date']}. Treat the survey date as the current date for this task.  
Answer only on the basis of information that would have been  
available to a person on that date.  
Do not reference or rely on any events or developments that occurred afterwards. Do not use knowledge of subsequent events to answer the question.  
Do not search for real data or base your answers on actual  
economic figures or official statistics. Instead, answer based on your own observations and general understanding of the economy at that time.

## Survey task

Answer the following questions about your expectations for gasoline prices, based on the specified survey date and the persona described above.

**A20. Five-year gasoline price expectations**

Do you think that the price of gasoline will go up during the next five years, will gasoline prices go down, or will they stay about the same as they are now?

Choose one answer:
- GO UP
- STAY THE SAME
- GO DOWN
- DON'T KNOW

**A20a. Expected five-year price change**

If your answer to A20 is GO UP or GO DOWN, answer the following:

About how many cents per gallon do you think gasoline prices will increase/decrease during the next five years compared to now?

Provide a whole-number estimate in cents per gallon. Report the magnitude as a positive number; the direction is already indicated by your answer to A20.


If you don't know the amount, answer DON'T KNOW.

If your answer to A20 is STAY THE SAME or DON'T KNOW, skip A20a and return null for this field.

**A20b. Twelve-month gasoline price expectations**

Now thinking only about the next twelve months, do you think that the price of gasoline will go up during the next twelve months, will gasoline prices go down, or will they stay about the same as they are now?

Choose one answer:
- GO UP
- STAY THE SAME
- GO DOWN
- DON'T KNOW

**A20c. Expected twelve-month price change**

If your answer to A20b is GO UP or GO DOWN, answer the following:

About how many cents per gallon do you think gasoline prices will increase/decrease during the next twelve months compared to now?

Provide a whole-number estimate in cents per gallon. Report the magnitude as a positive number; the direction is already indicated by your answer to A20b.


If you don't know the amount, answer DON'T KNOW.

If your answer to A20b is STAY THE SAME or DON'T KNOW, skip A20c and return null for this field.

## Output format

Return only a valid JSON object containing the four fields: A20, A20a, A20b, and A20c.

Follow these rules:

A20 and A20b: exactly one of GO UP, STAY THE SAME, GO DOWN, DON'T KNOW.

A20a and A20c must contain a positive integer representing the expected price change in cents per gallon.

Use 998 if the respondent is asked the question but answers DON'T KNOW.

Use null if the question is skipped according to the questionnaire's skip logic.

Do not include units in the numerical values.

Do not include explanations or any text outside the JSON object.
"""
````

````python
# Set OPENAI_API_KEY in your environment before running.
from getpass import getpass
from openai import OpenAI

api_key = getpass("Paste your API secret key here: ")
client = OpenAI(api_key=api_key)




response = client.responses.create(
    model="gpt-4.1",
    input=prompt
)

print(response.output_text)
````

````python
import json

output = response.output_text.strip()

# Remove Markdown code fences if present
if output.startswith("```"):
    output = output.removeprefix("```json").removeprefix("```")
    output = output.removesuffix("```").strip()

result = json.loads(output)

print(result)
````

````python
assert result["A20"] in [
    "GO UP", "STAY THE SAME", "GO DOWN", "DON'T KNOW"
]

assert result["A20b"] in [
    "GO UP", "STAY THE SAME", "GO DOWN", "DON'T KNOW"
]

for direction, magnitude in [
    ("A20", "A20a"),
    ("A20b", "A20c")
]:
    if result[direction] in ["STAY THE SAME", "DON'T KNOW"]:
        assert result[magnitude] is None
    else:
        assert (
            result[magnitude] == 998
            or (
                isinstance(result[magnitude], int)
                and not isinstance(result[magnitude], bool)
                and result[magnitude] > 0
            )
        )

print("Response passed validation!")
````
